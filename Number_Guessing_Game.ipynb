```python
import random
secret = random.randint(1,40)
max_attempts = 6
attempts = 0

print("--- Number Guessing Game ---")
while attempts < max_attempts:
  guess = int(input("Your guess (1-40): "))
  attempts +=1

  if guess == secret:
    print("Correct!", attempts, "tries.")
    break
  elif guess < secret:
    print("Too low.")
  else:
    print("Too high.")

else:
  print("Out of attempts! Answer:", secret)
```
